# Data Cleaning

DATA CLEANING PLAN
1. Drop duplicates
    - No missing ID rows can be dropped
    - if not duplicates will mess up my means
2. z-score standardization for all numerical variables
3. remove outliers after z-score filtering
    - what's the cutoff?
    - as NaN 
4. Removing NaNs


## Removing duplicates

```python
# Checking duplicates - different values from ProfileReport --> likely from treating NaNs in Cow ID as the same
df.duplicated().sum()
```

```python
# Keep all missing cow IDs, and drop duplicates when there is cow ID and all variables are exactly the same 
df_nodup = pd.concat([
    df[df["AnimalId"].notna()].drop_duplicates(keep="first"),
    df[df["AnimalId"].isna()]
], ignore_index=True)
```

```python
# Checking
df_nodup[df_nodup["AnimalId"].notna()].duplicated().sum()
```

## Z-score filtering

```python
# Z-score standardization
df_nodup["z_AvgMilkFlow"] = (df_nodup["AvgMilkFlow"] - df_nodup["AvgMilkFlow"].mean()) / df_nodup["AvgMilkFlow"].std()
df_nodup["z_Flow30_60Session"] = (df_nodup["Flow30_60Session"] - df_nodup["Flow30_60Session"].mean()) / df_nodup["Flow30_60Session"].std()
df_nodup["z_YieldSession"] = (df_nodup["YieldSession"] - df_nodup["YieldSession"].mean()) / df_nodup["YieldSession"].std()
```

```python
# Checking
df_nodup.head()
```

```python
# Outliers: more than 2 SDs away from the mean --> returned as Na
z_cols = ["z_AvgMilkFlow", "z_Flow30_60Session", "z_YieldSession"]

for col in z_cols:
    df_nodup.loc[df_nodup[col].abs() > 2, col] = np.nan
```

```python
# In rows with AnimalId, any z_score with NA is filtered out
has_id_clean = df_nodup[df_nodup["AnimalId"].notna()].dropna(subset=z_cols, how="any")
no_id = df_nodup[df_nodup["AnimalId"].isna()]

df_zscore = pd.concat([has_id_clean, no_id], ignore_index=True)
```

# Splitting Dataset

```python
# Step 1: Separate rows with missing CowID as the test set
test_df = df_zscore[df_zscore["AnimalId"].isna()]
```

```python
# Step 2: From the rest (CowID present), do stratified split by reproduction status
df_id = df_zscore[df_zscore["AnimalId"].notna()]

train_df, val_df = train_test_split(
    df_id,
    test_size=0.2,
    stratify=df_id["ReproductionStatus"],
    random_state=42
)
```
